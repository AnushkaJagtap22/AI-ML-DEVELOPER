# Remove HTML Tags

```python
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'',text)
```

```python
df['text'] = df['text'].apply(remove_html_tags)
```

# Remove URLs

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

```python
df['text'] = df['text'].apply(remove_url)
```

# Removing Punctuation

```python
import string,time
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punc(text):
    for char in exclude:
        text = text.replace(char,'')
    return text
```

```python
df['text'] = df['text'].apply(remove_punc)
```

# Chat word Treatment

```python
chat_words = {
    "u": "you",
    "r": "are",
    "ur": "your",
    "gr8": "great",
    "b4": "before",
    "l8r": "later",
    "btw": "by the way",
    "idk": "I do not know",
    "imo": "in my opinion",
    "omg": "oh my god",
    "lol": "laughing out loud",
    "brb": "be right back",
    "ttyl": "talk to you later",
    "thx": "thanks",
    "thnx": "thanks",
    "pls": "please",
    "plz": "please",
    "asap": "as soon as possible",
    "rn": "right now",
    "bc": "because",
    "cuz": "because",
    "msg": "message",
    "ppl": "people",
    "dm": "direct message",
    "tbh": "to be honest",
    "smh": "shaking my head",
    "tysm": "thank you so much",
    "np": "no problem",
    "omw": "on my way",
    "wanna": "want to",
    "gonna": "going to",
    "luv": "love",
    "k": "okay",
    "bff": "best friend forever"
}

def chat_conversion(text):
    new_text = []

    for word in text.split():
        clean_word = word.lower().strip(".,!?;:")

        if clean_word in chat_words:
            new_text.append(chat_words[clean_word])
        else:
            new_text.append(word)

    return " ".join(new_text)
```

```python
text = "OMG!!! u r gr8. Plz msg me ASAP, brb rn lol."

print(chat_conversion(text))
```

# Removing Stopwords

```python
from nltk.corpus import stopwords
```

```python
stop_words = set(stopwords.words('english'))
```

```python
def remove_stopwords(text):
    words = text.split()
    filtered_words = []

    for word in words:
        if word.lower() not in stop_words:
            filtered_words.append(word)

    return " ".join(filtered_words)
```

```python
text = "This product is very good and I am happy with the quality."

print("Original:", text)
print("After stopword removal:", remove_stopwords(text))
```

# Tokenization  -  is the process of splitting a piece of text into smaller units called tokens. These tokens can be words, sentences, or subwords.

## Using the split function

```python
# word tokenization
sent1 = "I am going to Delhi"
sent1.split()
```

```python
# sentance tokenization
sent2 = "I am going to Delhi. I will stay there for 2 days. I would love to visit India Gate"
sent2.split('.')
```

### Problems with the split() function

Punctuation remains attached to words, such as "Hello," and "amazing."

Multiple spaces are handled correctly by default, but the original spacing information is lost.

Contractions such as "don't" and "I've" are not split into meaningful components.

Punctuation, combined emojis, and hashtags remain grouped together.

## Using Regex

```python
text = "This phone costs 15000 rupees."

text = re.sub(r'\d+', '', text)
print(text)
```

# Stemming is a text preprocessing technique in Natural Language Processing (NLP) that reduces words to their root or stem by removing prefixes or suffixes.

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = ["playing", "played", "plays", "studies",
         "studying", "connected", "connection"]

for word in words:
    print(word, "->", ps.stem(word))
```

# Lemmatization is a Natural Language Processing (NLP) technique that converts a word into its base dictionary form, called a lemma, using vocabulary and grammatical information.

```python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

words = ["playing", "studies", "cars", "children", "running"]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```
