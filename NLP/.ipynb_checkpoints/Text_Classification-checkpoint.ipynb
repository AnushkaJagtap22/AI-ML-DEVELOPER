Text classification is a Natural Language Processing (NLP) task that assigns predefined categories or labels to text automatically.

For example, an email can be classified as Spam or Not Spam, and a customer review can be classified as Positive, Negative, or Neutral.

# Pipeline

Raw text  ->  Text preprocessing  ->  TF-IDF vectorization  ->    Machine learning classifier  ->  Predicted label: Positive / Negative

# Approaches to Text Classification in NLP

### 1. Heuristic-based approach

A heuristic approach uses predefined rules, keywords, or conditions to classify text. It does not require model training.

Example: Classifying messages as spam if they contain words such as “winner,” “free prize,” or “claim now.”

```python
def classify_text(text):
    spam_words = ["winner", "free prize", "claim now"]

    text = text.lower()

    for word in spam_words:
        if word in text:
            return "Spam"

    return "Not Spam"


print(classify_text("Congratulations! Claim now"))
print(classify_text("Hello, how are you?"))
```

### 2. API-based approach

An API-based approach sends text to an existing NLP service or pretrained model through an API. The service returns a classification or prediction.

Examples include sentiment analysis APIs, moderation APIs, and hosted transformer models.

# Generic example of calling an NLP classification API
import requests

url = "https://example.com/classify"

response = requests.post(
    url,
    json={"text": "This product is amazing"}
)

print(response.json())

### 3. Machine Learning (ML) approach

Traditional ML uses labeled training data to learn patterns from numerical text features. Text is commonly converted into features using Bag of Words or TF-IDF.

Popular algorithms include Naive Bayes, Logistic Regression, and Support Vector Machines (SVM).

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

texts = [
    "I love this product",
    "This is amazing",
    "I hate this product",
    "This is terrible"
]

labels = [
    "Positive",
    "Positive",
    "Negative",
    "Negative"
]

model = make_pipeline(
    TfidfVectorizer(),
    MultinomialNB()
)

model.fit(texts, labels)

print(model.predict(["This product is amazing"]))
```

### 4. Deep Learning (DL) approach

Deep learning uses neural networks to learn complex representations and patterns from text. Models may use embeddings, recurrent neural networks (RNNs), LSTMs, CNNs, or Transformers such as BERT.

Example using a pretrained Hugging Face sentiment classifier:

```python
# Install first:
# pip install transformers torch

from transformers import pipeline

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

result = classifier("This product is absolutely amazing!")
print(result)
```
