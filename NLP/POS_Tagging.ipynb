POS (Part-of-Speech) Tagging is an NLP technique that identifies the grammatical category of each word in a sentence, such as a noun, verb, adjective, or adverb.

Example:

Sentence: The beautiful girl sings sweetly.

Word	POS tag	Meaning

The	DET	Determiner

beautiful	ADJ	Adjective

girl	NOUN	Noun

sings	VERB	Verb

sweetly	ADV	Adverb

```python
import nltk

nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")

text = "The beautiful girl sings sweetly."

# Tokenize the sentence
words = nltk.word_tokenize(text)

# Perform POS tagging
tags = nltk.pos_tag(words)

print(tags)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "The beautiful girl sings sweetly."
doc = nlp(text)

for token in doc:
    print(token.text, token.pos_, token.tag_)
```
