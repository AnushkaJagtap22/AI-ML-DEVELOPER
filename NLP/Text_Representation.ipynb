# Text representation 

Text representation is the process of converting text data into numerical form so that machine learning algorithms can understand and process it.

# Types of text representation

1. Bag of Words (BoW)
2. TF-IDF
3. Word Embeddings
4. Contextual Representations


# Common NLP terms

Corpus - A collection of text documents - A dataset of product reviews

Vocabulary - All unique words in a corpus - ['good', 'product', 'bad']

Token - A word or text unit - "amazing"

Frequency - Number of times a word occurs - good: 5

# One Hot Encoding

One-hot encoding is a technique used to represent categorical data or words as binary vectors containing only 0s and a single 1.

Suppose the vocabulary is:

["apple", "banana", "mango"]

The one-hot representation is:

Word	Apple	Banana	Mango

apple	1	0	0

banana	0	1	0

mango	0	0	1

```python
from sklearn.preprocessing import OneHotEncoder

words = [["apple"], ["banana"], ["mango"], ["apple"]]

encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(words)

print("Categories:", encoder.categories_[0])
print(encoded)
```

# Bag Of Words

Bag of Words (BoW) is a text representation technique in Natural Language Processing (NLP) that converts text into numerical vectors based on the frequency of words in a document.

## Example

Consider these three sentences:

- Document 1: I love Python
- Document 2: I love NLP
- Document 3: Python is amazing

### Step 1: Create the vocabulary

Unique words: ['I', 'love', 'Python', 'NLP', 'is', 'amazing']

### Step 2: Count each word

Document	I	love	Python	NLP	is	amazing

Document 1	1	1	1	0	0	0

Document 2	1	1	0	1	0	0

Document 3	0	0	1	0	1	1

Each row is the Bag of Words vector for a document.

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "I love Python",
    "I love NLP",
    "Python is amazing"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

print("Vocabulary:", vectorizer.get_feature_names_out())
print("BoW Matrix:\n", X.toarray())
```

# Tf-Idf

TF-IDF (Term Frequency–Inverse Document Frequency) is a text representation technique used in Natural Language Processing (NLP) to convert text into numerical values. It gives higher importance to words that are frequent in a particular document but less common across the entire collection of documents.

Suppose we have these three documents:

- Document 1: I love Python
- Document 2: I love NLP
- Document 3: Python is amazing

The word love appears in two documents, while amazing appears in only one. Therefore, amazing gets a higher IDF score than love, because it is rarer across the collection.

TF-IDF combines this rarity with how often the word appears in each document to calculate its weight.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "I love Python",
    "I love NLP",
    "Python is amazing"
]

# Create the TF-IDF vectorizer
vectorizer = TfidfVectorizer()

# Convert text into TF-IDF vectors
X = vectorizer.fit_transform(corpus)

# Display vocabulary
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

# Display TF-IDF matrix
print("\nTF-IDF Matrix:")
print(X.toarray())
```

# Word2Vec

Word2Vec is a word embedding technique in Natural Language Processing (NLP) that converts words into dense numerical vectors. Unlike Bag of Words and TF-IDF, Word2Vec can learn semantic relationships between words based on their surrounding context.

### How Word2Vec works

Word2Vec learns word representations using a neural network and two main architectures:

1. CBOW (Continuous Bag of Words)

Predicts a target word using its surrounding context words.

Example: “I ___ Python” → predict “love”

2. Skip-gram

Predicts surrounding context words using a target word.

Example: “love” → predict “I” and “Python”

```python
from gensim.models import Word2Vec

# Sample training data
sentences = [
    ["i", "love", "python"],
    ["i", "love", "machine", "learning"],
    ["python", "is", "a", "programming", "language"],
    ["machine", "learning", "is", "interesting"],
    ["i", "enjoy", "coding", "in", "python"]
]

# Train the Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=3,
    min_count=1,
    sg=1,
    workers=1,
    seed=42,
    epochs=100
)

# Display the vector for a word
print("Vector for Python:")
print(model.wv["python"])

# Find similar words
print("\nSimilar words:")
print(model.wv.most_similar("python", topn=3))
```
